```python
# Combining arrays
# concatenate()

a = np.array([
    [1, 2],
    [3, 4]
])

b = np.array([
    [5, 6],
    [7, 8]
])

c = np.concatenate([a,b], axis=0) # add rows
print(c)
d = np.concatenate([a,b], axis=1) # adds columns
print(d)
print('-'*30)
e = np.vstack((a,b)) # adds as row
print(e)
f = np.hstack((a,b)) # adds as columns
print(f)
print('-'*30)
```

```python
# Filtering
# Selecting elements from an array based on a condition
# NumPy uses C-style logical operators

# With flat lists
ages = np.array([
    [21, 17, 19, 20, 16, 30, 18, 65],
    [39, 22, 15, 99, 18, 19, 20, 21]
])

teens = ages[ages < 18]
print(teens)

adults = ages[(ages >= 18) & (ages < 65)]
print(adults)

seniors = ages[ages >= 65]
print(seniors)


# Preserve original structure
# np.where(condition, value_if_true, value_if_false)

adults = np.where(ages >= 18, ages, np.nan)
print(adults)
```

```python
# Random Numbers

# Seed lets you create the same random numbers repeatedly

# Integers
rng = np.random.default_rng(seed=1)

print(rng.integers(1, 10))
print(rng.integers(low=1, high=10))
print(rng.integers(low=1, high=10, size=(2, 2)))
print(rng.integers(low=1, high=10, size=(3, 4)))


# Floating-point numbers
np.random.seed(seed=1)

print(np.random.uniform())
print(np.random.uniform(-1, 2))
print(np.random.uniform(low=-1, high=2))
print(np.random.uniform(low=-1, high=2, size=3))
print(np.random.uniform(low=-1, high=2, size=(3, 2)))
```

```python

# Shuffling numbers
rng = np.random.default_rng()

ar = np.array([1, 2, 3, 4, 5])

print(ar)

rng.shuffle(ar)

print(ar)


# Random choosing
rng = np.random.default_rng()

alphabets = ['A', 'B', 'C', 'D', 'E', 'F', 'G', 'H']

alphabet = rng.choice(alphabets)
print(alphabet)

alps = rng.choice(alphabets, size=3)
print(f'\n{alps}')

alps = rng.choice(alphabets, size=(2, 3))
print(f'\n{alps}')
```
