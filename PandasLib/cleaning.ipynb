##### replace()

```python
df['Address'] = df['Address'].replace('N/a', '')
```

```python
# replacing values

df['Do_Not_Contact'] = df['Do_Not_Contact'].replace(['Y', 'YES'], 'Yes')
df['Paying Customer'] = df['Paying Customer'].replace(['Y', 'YES'], 'Yes')

df['Do_Not_Contact'] = df['Do_Not_Contact'].replace(['n', 'NO', 'no'], 'No')
df['Paying Customer'] = df['Paying Customer'].replace(['n', 'NO', 'no'], 'No')

# or 
''' 
df['Do_Not_Contact'] = df['Do_Not_Contact'].replace({
    'Y': 'Yes',
    'YES': 'Yes',
    'n': 'No',
    'NO': 'No',
    'no': 'No'
})
'''
# best
''' 
df['Do_Not_Contact'] = (
    df['Do_Not_Contact']
    .str.strip()
    .str.lower()
    .replace({'y': 'Yes', 'yes': 'Yes', 'n': 'No', 'no': 'No'})
)
'''
```

```python
# Problem: If do not call is -> Yes, then remove their phone number

# Method 1
# row by row method
for x in df.index:
    if df.loc[x, "Do_Not_Contact"] == 'Yes' or df.loc[x, "Do_Not_Contact"] == '':
        df.loc[x, 'Phone_Number'] = pd.NA

df.head(10)

# Method 2
# pandas's method
# df.loc[condition, column] = value
df.loc[df['Do_Not_Contact'] == 'Yes'  ,  'Phone_Number'] = pd.NA

df.loc[
    (df['Do_Not_Contact'] == 'Yes') | (df['Do_Not_Contact'] == ''),
    'Phone_Number'
] = pd.NA
```
