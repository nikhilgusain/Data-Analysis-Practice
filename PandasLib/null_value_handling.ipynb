````python
df['society'] = df['society'].fillna('Unknown')
````

````python
# Numeric columns
# can be filled with mean or median

df['bath'] = df['bath'].fillna(df['bath'].mean())

# For variables such as number of bathrooms/balconies, 
# median is often more sensible: 

df['bath'] = df['bath'].fillna(df['bath'].median())
````

#### Dropping data

##### dropna()  
axis = 1 for droping columns default is 0(rows)  
##### thresh
minimum no. of columns that need to be empty to drop a row.  
df.dropna(thresh = 2).
keeps rows with >=2 non-null values  
  
##### For columns:
  
df.dropna(axis=1, thresh=2)  
  
→ keep columns with ≥ 2 non-null values.  

##### Very important distinction
df.dropna(thresh=2)  

does NOT mean "allow 2 missing values."  

It means:  

Require at least 2 valid/non-null values.  
  
For example, with 5 columns:  
  
thresh=2 → at least 2 non-NaN values  
thresh=3 → at least 3 non-NaN values  
thresh=5 → all 5 must be non-NaN  

####  group-wise imputation


Group-wise Imputation  

Group-wise imputation means filling missing values  
using information from the same group   
instead of using one overall statistic.
