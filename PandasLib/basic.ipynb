## Series

```python
data = [100, 200, 300 ]
s = pd.Series(data)
print(s)

# manual labels
s2 = pd.Series(data, index=['a', 'b', 'c'])
print(s2)
```

```python
# loc
# loc - locating by lable
print(s2.loc['a'])
print(s2.loc['b'])

# iloc
# integer locating
print('-'*30)
print(s2.iloc[0])
print(s2.iloc[2])
```

```python
data = [100, 102, 104, 200, 202]
s = pd.Series(data, index = ['a', 'b', 'c', 'd', 'e'])
print(s[s>=200])
```

#### Filtering

```python
df.columns
```

```python
print(df[df['Defense']>50])
```

```python
df[(df['HP'] > 80) & (df['Attack'] > 50)]
```

```python
df[(df['HP']>80) | (df['Attack'] >80)]
```

```python
#  filter + specific columns

# df.loc[condtions, ['col1' , 'col2' , 'col3']]

df.loc[
    (df['HP'] > 100) | (df['Attack'] > 80),
    ['Name', 'HP', 'Attack']
]

df.loc[
    (df['Type 1'] == 'Grass') | (df['Type 2'] == 'Grass'),
    ['Name', 'Type 1', 'Type 2']
]
```

```python
df.query("HP > 80 and Attack < 50 ")
```

```python
df.query("`Type 1` == 'Grass' or `Type 1` == 'Water'")[['Name', 'Type 1']]
```

```python
# isin 
# chekcs is a values is present in a column or  no 
# return boolean val.

print(df['Type 1'].isin(['Grass', 'Water']))

# no in 
df[~df['Type 1'].isin(['Grass'])]
```

```python
# between
# .between(x, y) x and y are inclusive >=, <=
df['HP'].between(50, 100)

df.loc[
    (df['HP'].between(50,100)),
    ['Name', 'HP']
]

#  > HP <
df['HP'].between(50, 100, inclusive='neither')

#  50 < HP <= 100
df['HP'].between(50, 100, inclusive='right')

# 50 <= HP < 100
df['HP'].between(50, 100, inclusive='left')
```

##### Adding , Droping , Modifying columns

```python
# adding

df['Power'] = df['Attack'] + df['Defense']

# Modifying values
df['HP'] = df['HP'] * 2
df['HP'] = df['HP'] // 2
```

```python
# Renameing 
# renaming return new df, use inplace for changing in df
df.rename(columns={'HP':'Health'}, inplace=True)
df.rename(columns=
          {'Health' : 'HP',
           'Power':'strength'},
           inplace=True)

df2 = df
```

```python
# droping, return new df, use inplace 
df.drop(columns=['strength', 'HP'], inplace= True)
df = df2
```

#### Aggregation
taking mulitple values to get summa

##### Basics

```python
# Basic agrregation
print(df['HP'].sum())


print(df['HP'].min())
print(df['HP'].max())

print(df['HP'].mean())
print(df['HP'].mode())
print(df['HP'].median())


print(df['HP'].count())  # non null values

print(df['HP'].std())
print(df['HP'].var())
```

```python
# agrregation on whole data frame

print(df.sum(numeric_only=True))
print('-'*40)
print(df.mean(numeric_only=True))
```

```python
# multiple agg
df['HP'].agg(['sum', 'mean', 'median', 'std', 'var'])
```

##### group_by()

```python
# employee data set 
edf = pd.read_csv('emp.csv')
print(edf.columns)
# edf.head()
```

```python
df.groupby('Type 1')['Attack'].mean()
```

```python
edf.groupby('did')['SALARY'].sum()
```

```python
print(edf.groupby('did')['SALARY'].mean())
print(edf.groupby('did')['SALARY'].sum())
print(edf.groupby('did')['SALARY'].min())
print(edf.groupby('did')['SALARY'].max())
print(edf.groupby('did')['SALARY'].count())
```

```python
# counting
# count employees in each dep
print(edf.groupby('did')["EMPLOYEE_ID"].count().sort_values(ascending=False))

print(edf.groupby('did').size())
```

```python
# Creating group

group = edf.groupby('did')

# Note: count; tells  non-null values, size tells total rows
group.size().sort_values()
```

```python
df.groupby('Type 1').size()  # similar to value_count()
```

```python
# single column + multiple agg.  + sort
# Note : df.groupby('column')['column2'].aggregation().sort_values()
df.groupby('Type 1')['Attack'].agg(['sum', 'mean', 'min', 'max']).sort_values('mean')
```

```python
# multiple columns + single agg.

df.groupby('Type 1')[['Attack', 'Speed', 'Defense']].mean().sort_values('Attack')
```

```python
group['SALARY'].mean().sort_values(ascending=False)
```

```python
# Multiple statistics + multiple columns
df.groupby('Type 1').agg(
    {
        'Attack':'mean',
        'Attack':'sum',
        'Speed':'min',
        'HP':'std'
    }
)
```
